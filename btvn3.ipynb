```python
# W3A16
a = float(input())
n = int(a)
if a!=n:
    san = n if a>0 else n-1
    tran = n+1 if a>0 else n
    if a>0:
        gan = n+1 if a-n >= 0.5 else n
    else:
        gan = n-1 if a-n <= -0.5 else n
else:
    san = tran = gan = n
print(tran, san, gan)
```
