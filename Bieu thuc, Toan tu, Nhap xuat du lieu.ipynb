```python
# W2A3 [ModAndDiv?]
a,b = map(int, input("NHập 2 số nguyên: ").split())
print("Tổng =", a+b)
print("Hiệu =", a-b)
print("Tích =", a*b)
print("Thương nguyên =", a//b)
print("Dư =", a%b)
print("Thương thực =",f"{a/b:.2f}")
```

```python
# W2A4 [trung binh cua 6]
a1,b1,c1,a2,b2,a3 = map(float,input("Nhập 6 đầu điểm: ").split())
tb = ((a1+b1+c1) + (a2+b2)*2 + a3*3)/10
print(f"{tb:.1f}")
```
