```python
# bai3
import math
a,b,c = sorted(map(float, input().split()))
if a>0 and a+b>c:
    vuong = math.isclose(a**2 + b**2, c**2)
    can = a==b or b==c
    deu = a==c
    cv = a+b+c
    p = cv/2
    dt = round(math.sqrt(p*(p-a)*(p-b)*(p-c)), 3)
    if deu:
        loai = "tam giác đều"
    elif can and vuong:
        loai = "tam giác vuông cân"
    elif vuong:
        loai = "tam giác vuông"
    elif can:
        loai = "tam giác cân"
    else:
        loai = "tam giác thường"
    print(f"3 cạnh lập thành {loai} có chu vi {cv} và diện tích {dt}")
else:
    print("không phải tam giác")
```

```python
# bai5
import math
a,b,c = map(float, input().split())
d = b**2 - 4*a*c
if d<0:
    print("vô nghiệm")
elif d>0:
    nghiem = [(-b+math.sqrt(d))/2*a, (-b-math.sqrt(d))/2*a]
    print("Phương trình có nghiệm: x =", end=" ")
    print(*nghiem, sep="; ")
elif d==0:
    nghiem = -b/a
    print(f"Phương trình có nghiệm kép: x = {nghiem}")
```

```python
# bai8
a,b,m,c,d,n = map(float, input().split())
xt = (d*m-b*n)
yt = (a*n-c*m)
mau = (d*a-b*c)
if mau==0:
    if xt==0 and yt==0:
        print("Hệ phương trình vô số nghiệm")
    elif xt!=0 or yt!=0:
        print("Hệ phương trình vô nghiệm")
else:
    print(f"Hệ phương trình có nghiệm: (x; y) = ({xt/mau}; {yt/mau})")
```

```python
# bai10
import math
x,y,r,a,b = map(int, input().split())
d = math.sqrt((a-x)**2 + (b-y)**2)
if r>0:
    if r>d:
        loai = "nằm trong"
    elif r==d:
        loai = "thuộc"
    elif r<d:
        loai = "nằm ngoài"
    print(f"Điểm A({a}; {b}) {loai} đường tròn tâm I({x}; {y}) bán kính {r}")
else:
    print("??")
```
